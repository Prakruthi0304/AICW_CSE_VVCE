```python
# __init__ constructor -used to initialize

class person :
    def __init__(self , name , age , school):
        self.name=name   # instance variable 
        self.age=age
        self.school=school
    def myfun(self):
        print("heyy , this is ",self.name)
        print("hmmm age is probably", self.age)
        print("met at ", self.school)
```

```python
p3=person ('gg',23,'lib')
p3.myfun()
```

```python
# accessing attributes and method

class emp:
    name='GP'
    salary=10
    def show(self):
        print(self.name)
        print(self.salary)
```

```python
e1=emp()
```

```python
# using get attribute insted e1.name to access the value 

print(getattr(e1,'name'))
```

```python
# has attribute - return true if  object has attribute or false 

print(hasattr(e1,'salary'))
```

```python
print(hasattr(e1,'gen'))
```

```python
# set attribute - sets an attribute 

(setattr(e1,'gen','F'))
```

```python
print(getattr(e1,'gen'))
```

```python
# delete attribute -del the attribute 

delattr(e1,'gen')
```

```python
print(getattr(e1,'gen'))  # we deleted gen 
```

```python
class S:
    def __init__(self,USN , name , sec): # constructor method 
        # instance variable 

        self.USN=USN
        self.name=name
        self.sec=sec

        # instance method -to access instance variable 

    def dis(self):
            print('USN ',self.USN ,'name ',self.name , 'sec ',self.sec)

        #instance method to modify the instance variable 
    def update(self,USN,sec):
            self.USN=USN
            self.sec=sec
```

```python
p6=S('4VV24CS099','GP','F')
p6.dis()
```

```python
p6.update("4VV24CS091",'E')
p6.dis()
```
