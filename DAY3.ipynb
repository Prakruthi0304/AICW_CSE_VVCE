# class method decorator


```python
from datetime import date 
```

```python
class student:
    def __init__(self , name , age):
        self.name=name 
        self.age=age 

    @classmethod 
    def calculate_age(cls , name , birthyear):
        return cls(name,date.today().year-birthyear)

    def show(self):
        print(self.name + "'s age is '"+ str(self.age))
```

```python
age=student('gp',20)
age.show()
```

```python
by=student.calculate_age('gp',2006)
by.show()
```

```python
# class method  
class School:
    name="st joseph"
    @classmethod
    def school_name(cls):
        print('school name is ',cls.name)
    School.school_name=classmethod(School.school_name)

School.school_name()
```

```python
# static method 

class emp:
    @staticmethod 
    def sample(x):
        print('inside static method ',x)
emp.sample(10)

emp1=emp()
emp1.sample(10)
```

```python
class BankAccount:
    def __init__(self,name,balance=0):
        #instance attributes
        self.name=name
        self.balance=balance
    #deposit money
    def deposit(self,amount):
        if amount>0:
            self.balance+=amount
            print("Deposit Successfully!")
            print("Deposited:",amount)
        else:
            print("Invalid deposit amount")
    #withdaw money
    def withdraw(self,amount):
        if amount <=0:
            print('Invalid withdrawl amount!')
        elif amount > self.balance:
            print('Insufficient balance!')
        else:
            self.balance -=amount
            print("Withdrawal successfull")
            print("Withdrawn:",amount)
    #check balance
    def check_balance(self):
        print("Current balance:",self.balance)
    #display account details
    def display_details(self):
        print("Account Holder:",self.name)
        print("Balance:",self.balance)
        
```

```python
acc1=BankAccount('GP',10000)
acc1.display_details()
```

```python
acc1.deposit(3999)
```

```python
acc1.check_balance()
```

```python
acc1.withdraw(477)
```

```python
acc1.check_balance()
```

```python
# parent class
class Person:
    def person_info(self,name,age):
        print('invalid person class')
        print('name:',name , 'age:',age)
# parent 2
class Company:
    def company_info(self,company_name,location):
        print('inside the company class')
        print('name:',company_name,'location:',location)
```

```python
# child class
class Employee(Person , Company):
    def Employee_info(self,salary,skill):
        print('inside employee class-child class')
        print('salary:',salary,'Skills:',skill)
```

```python
# create an  object of employee
emp2=Employee()
emp2.person_info('GP',20)
emp2.company_info('edunet','bengalore')
emp2.Employee_info(30000,'prompt engineer')
```

```python
#public member

class Employee:
    def __init__(self,name , salary):
         self.name=name 
         self.salary=salary
   
   
    # public instance method 

    def show(self):
        # access public data members 

        print('name',self.name , 'salary',self.salary)
```

```python
# private member
emp =Employee('GP',10000)
# access the public dat members 
print('name',emp.name,'salary',emp.salary)
emp.show()
```

```python
# create  an object 
emp=Employee('GP',5000)
# acces the private data member 

print("salary",emp.__salary)
```

```python
# protected memeber 
class Company:
    def __init__(self):
        # protected member 
        self._project='NLP'
# child class
class Employee(Company):
    def __init__(self,name):
        self.name=name
        Company.__init__(self)
    def show(self):
        print("employee name :",self.name)
        # access the protected member 
        print('working on project',self._project)
```

```python
c=Employee('GP')
c.show()
```

```python
# direct access
print('project',c._project)
```

```python
class India():
    def capital(self):
        print("new delhi is the capital of india")
    def lang(self):
        print("india has 22 official langauges")
    def type(self):
        print('india is a developing nation!')

class USA():
    def capital(self):
        print("Washington is the capital of USA")
    def lang(self):
        print("english is the official lang")
    def type(self):
        print('USA is a developed nation!')
```

```python
ob_in=India()
ob_usa=USA()
for country in (ob_in,ob_usa):
    country.capital()
    country.lang()
    country.type()
```

```python

#polymorphism with inheritance
class Bird:
    def intro(self):
        print('There are many types of birds!')
    def flight(self):
        print('Most of the birds can fly but some cannot!')
class sparrow(Bird):
    def flight(self):
        print("Sparrows can fly")
class ostrich(Bird):
    def flight(self):
         print("Ostrich cannot fly")
   
```

```python
bi1=Bird()
bi1.intro()
bi1.flight()
```

```python
#overloading
class complex:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def __add__(self,other):
        return self.a+other.a,self.b+other.b
ob1=complex(1,2)
ob2=complex(2,3)
ob3=ob1+ob2
print(ob3)
```
